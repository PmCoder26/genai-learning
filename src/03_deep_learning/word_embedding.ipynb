# Word Embeddings

An embedding is a dense numerical representation of a word. Instead of a very large sparse one-hot vector, each word is mapped to a smaller vector whose values are learned from context. Words with similar meanings tend to have similar embedding vectors.

This notebook demonstrates the basic idea using a small text dataset and a Keras Embedding layer. The goal is to understand how a word index is converted into a learned dense vector representation.


```python
# One-hot encoding is a simple way to represent words numerically.
# Each word gets a unique index in the vocabulary, and a long vector is created

# where only one position is 1 and all others are 0.from tensorflow.keras.preprocessing.text import one_hot

# This is useful for understanding the idea, but it is not efficient for NLP models,# because words are represented as very large sparse vectors.
```

```python
# An embedding layer is a learnable lookup table.
# Instead of using huge one-hot representations, every word index is mapped to a
# compact dense vector of fixed size. These vectors are learned during training, and
# similar words tend to end up close together in embedding space.
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.text import one_hot
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
# These example sentences are a small text corpus for demonstration purposes.
# The goal is to show how text can be converted into numeric form before passing
# it into a neural network. Even in a tiny example, we can see repeated terms and
# similar sentence structures that learning-based representations can capture.
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good'
    ]
    
```

```python
# The vocabulary size is the total number of unique words the model can represent.
# A large value like 10000 makes sure that the sample text fits comfortably
# inside the vocabulary, even if the dataset grows later.
vocab_size = 10000
```

```python
# Each sentence is turned into a list of one-hot encodings for its words.
# This creates a numeric representation that a model can accept, although it is
# still high-dimensional and sparse compared with learned embeddings.

one_hot_repr = [one_hot(words, vocab_size) for words in sent]
```

```python
# Neural networks usually expect all input sequences to have the same length.
# Padding adds zero values at the start of shorter sequences so the data can be
# stacked into a consistent matrix. Here, each sentence is padded to a length of 10.
sent_len = 10

embedded_docs = pad_sequences(one_hot_repr, padding='pre', maxlen=sent_len)
```

```python
# The embedding dimension controls how large each learned word vector is.
# A dimension of 10 means each word is represented by a 10-value vector, rather

# than by a huge one-hot vector. This is a very small example, but it shows the idea.
dimension = 10
```

```python
# We create a very small neural network that contains only an Embedding layer.
# The layer receives integer word indices and converts each one to a dense vector.
# Keras handles the mapping automatically, and the model can be trained later.
model = Sequential()

model.add(Embedding(vocab_size, dimension, input_length=sent_len))

model.compile('adam', 'mse')
```

```python
model.summary() 
# shows the architecture and parameter counts.
# This helps us understand how the embedding layer maps the vocabulary into a

# smaller vector space. In a larger model, this layer would learn meaningfulmodel.summary()
# semantic patterns from the data during training.
```

```python
# The model generates dense vectors for each word position in the input sequences.
# These learned vectors are the actual embedding output. In a real application,

# these vectors are usually used as input features for downstream tasks such asmodel.predict(embedded_docs)
# classification, sentiment analysis, or retrieval.
```

```python
# This example is intentionally small to make the concept easy to understand. In real NLP
# projects, embeddings are trained on large corpora and capture richer relationships

# between words, phrases, and contexts. The core idea is the same: transform discrete
# tokens into dense vectors that encode meaning in a machine-readable form.
```
